# Exercise 2.C Lab 3 - Tax Functions
This notebook demonstrates custom functions that calculate payroll tax withholdings, including Social Security, Medicare, and federal income tax based on withholding codes.

def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        # code 4+ .. 18% minus 0.5% for each withholding over 4
        rate = 0.18 - ((withholding_code - 4) * 0.005)
    return gross_pay * rate
```

```python
gross = 750
code = 0
print(f"Person 1 | Gross: ${gross:.2f} | Withholding Code: {code}")
print(f"Social Security:   ${get_soc_sec_tax(gross):.2f}")
print(f"  Medicare:        ${get_medicare_tax(gross):.2f}")
print(f"  Federal Tax:     ${get_federal_tax(gross, code):.2f}")
```
