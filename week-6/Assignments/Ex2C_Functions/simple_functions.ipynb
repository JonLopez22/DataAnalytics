# Exercise 2.C Creating and calling functions
## Lab 1. Defining custom functions

```python
# Favorite things
def favorite_things(name, movie, music):
    return (f"Name: {name}\n"
            f"Favorite movie: {movie}\n"
            f"Favorite musician/band: {music}")
print(favorite_things('Jonathan', 'The Dark Knight', 'Drake'))
```

```python
# why im here
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Year Up United Data Analyst program to build a career in tech and grow my skills in data."))
```

```python
# Favorite place
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place('New York City', 'there is always something to do'))
```

```python
print(favorite_place('the Dominican Republic', 'the beaches are so amazing and beautiful'))
```
