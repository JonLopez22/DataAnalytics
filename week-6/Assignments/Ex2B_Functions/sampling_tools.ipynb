```python
import random
```

```python
products = ['Laptop', 'Monitor', 'Keyboard', 'Mouse', 'Webcam', 'Headset', 'Docking Station', 'USB Hub', 'Desk Lamp', 'Surge Protector']
```

```python
# a)
daily_pick = random.choice(products)
print(f"Product of the day: {daily_pick}")
```

```python
# b)
survey_picks = random.sample(products, 3)
print(f"Survey products: {survey_picks}")
```

```python
# c)
random.shuffle(products)
print(f"Shuffled products: {products}")
```

```python
# d)
transactions = random.randint(50, 300)
print(f"Daily transaction count: {transactions}")
```
