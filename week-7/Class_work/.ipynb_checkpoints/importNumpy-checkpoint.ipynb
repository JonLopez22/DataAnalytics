```python
import numpy as np
```

```python
# 1D Array

arr1 = np.array([1, 2, 3])

print(arr1)
print(f"Shape: {arr1.shape}")
print(f"Dimensions: {arr1.ndim}")
```

```python
# 2D array
arr2 = np.array([
    [1, 2, 3],
    [1, 2, 3],
    [1, 2, 3]
])

print(arr2)
print(f"Shape: {arr2.shape}")
print(f"Dimensions: {arr2.ndim}")
```

```python
# 3D Array

arr3 = np.array([
    [[1, 2, 3], [1, 2, 3], [1, 2, 3], [1, 2, 3]],
    [[1, 2, 3], [1, 2, 3], [1, 2, 3], [1, 2, 3]],
    [[1, 2, 3], [1, 2, 3], [1, 2, 3], [1, 2, 3]]
])

print(arr3)
print(f"Shape: {arr3.shape}")
print(f" Dimensions: {arr3.ndim}")
```

```python
# Arrays
print("=== NumPy Array Creations Functions ===\n")

# 1 array()
a = np.array([1, 2, 3, 4, 5])
print("np.array([1,2,3,4,5])")
print(a, "\n")
```

```python
# 2 zeros() - array of zeros
b = np.zeros((2, 3))
print("np.zeros((2,3))")
print(b, "\n")
```

```python
# 3  ones() - array of ones
c = np.ones((2, 2))
print("np.ones((2,2))")
print(c, "\n")
```

```python
# 4 full() - array filled with a constant value
d = np.full((3, 2), 7)
print("np.full((3,2), 7)")
print(d, "\n")
```

```python
# Arrays
a = np.array([10, 20, 30, 40])
print(a * 2)
print(a.mean())
print(a.dtype)
```

```python
# Create 2D NumPu array
a = np.array([[10, 20, 30],
             [40, 50, 60]])
print("Array:")
print(a)

# Shape
print("\nShape of array:", a.shape)

# .ndim
print("Number of dimensions (ndim):", a.ndim)

# .dtype - Data Type
print("Data type (dtype):", a.dtype)

# .size
print("total number of elements:", a.size)
```
