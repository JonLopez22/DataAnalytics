```python
import pandas as pd

data = {
    "Student":["Amy", "Bob", "Cara", "Dan", "Eva"],
    "Score":[85,92,78,95,100]
}

df = pd.DataFrame(data)
print(df)
print()
```

```python
print("SUM:", df["Score"].sum())
print("COUNT:", df["Score"].count())
print("MEAN:", df["Score"].mean())
print("MEDIAN:", df["Score"].median())
print("MIN:", df["Score"].min())
print("MAX:", df["Score"].max())
```

```python
print("\nAggregate Results:")
print(df["Score"].agg(["sum","mean","median","min","max"]))
```

```python
# Pandas GroupBy

data = {
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Employee": ["Amy", "Bob", "Cara", "Dan", "Eva", "Frank"],
    "Salary": [70000, 50000, 80000, 60000, 55000, 65000]
}
df = pd.DataFrame(data)
print(df)
print()
```

```python
# Group by department
grouped = df.groupby("Department")
print("SUM")
print(grouped["Salary"].sum())
```

```python
print("\nMean")
print(grouped["Salary"].mean())

print("\nCount")
print(grouped["Employee"].count())
```

```python
df = pd.DataFrame({
    "Student": ["Amy", "Bob", "Cara", "Dan"],
    "Score": [88, 95, 79, 91]
}, index=[3,1,2,0])
print("ORIGINAL DATAFRAME")
print(df)
```

```python
print("\nSORTED INDEX")
print(df.sort_index())
```

```python
df["Rank"] = df["Score"].rank(ascending=False)
print("\nWITH RANKINGS")
print(df)
```

```python
# Descending
df = pd.DataFrame({
    "Student": ["Amy", "Bob", "Cara", "Dan"],
    "Score": [88, 95, 79, 91]
}, index=[3, 1, 2, 0])

print("ORIGINAL DATAFRAME")
print(df)
```

```python
print("\nSORTED INDEX (Highest to Lowest)")
sorted_df = df.sort_index(ascending=False)
print(sorted_df)
```

```python
df["Rank"] = df["Score"].rank(ascending=False)
ranked_df = df.sort_values(by="Rank")

print("\nRANKINGS (Highest to Lowest)")
print(ranked_df)
```

```python
#DataFrame 1: Students
students = pd.DataFrame({
    "StudentID": [1, 2, 3, 5],
    "Score": [85, 92, 78, 88]
})
scores = pd.DataFrame({
    "StudentID": [1, 2, 3, 5],
    "Score": [85, 92, 78, 88]
})


print("STUDENTS DATAFRAME")
print(students)
print("\nSCORES DATAFRAME")
print(scores)

print()
```

```python
# JOINS
# 1. INNER JOIN
inner_merge = pd.merge(students, scores, on="StudentID", how="inner")
print("\nINNER MERGE (Only matching StudentID)")
print(inner_merge)
```

```python
# 2. LEFT JOIN
left_merge = pd.merge(students, scores, on="StudentID", how="left")
print("\nLEFT MERGE (all students kept)")
print(left_merge)
```

```python
# 3. RIGHT JOIN
right_merge = pd.merge(students, scores, on="StudentID", how="right")
print("\nRIGHT MERGE (all scores kept)")
print(right_merge)
```

```python
# 4. OUTER JOIN
outer_merge = pd.merge(students, scores, on="StudentID", how="outer")
print("\nOUTER MERGE (all data from both)")
print(outer_merge)
```
