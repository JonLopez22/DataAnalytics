## series

```python
# Four plain objects
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
# Convert each to a Pandas series
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac))
ser4 = pd.Series(ad)

print ('ser1:\n', ser1)
print('\nser2:\n', ser2)
print('\nser3\n', ser3)
print('\nser4\n', ser4)
```

```python
# Creating a series from literal values

ser5 = pd.Series(17, 18, 19, 20)

# TypeError - Series() got multiple positional arguments,
# you must pass values as a list, can't pass it as seperate arguments. 
```

## DataFrames from Series

```python
# Creating df1 from ser1

df1 = pd.DataFrame(ser1)
df1
```

```python
# Adding ser2 and ser3 as new columns using insert() function

df1.insert(1, 'col2', ser2)
df1.insert(2, 'col3', ser3)
df1
```

```python
# Add ser4 as a third column (new)

df1['col4'] = ser4

df1
# ser4 was created from a dictionary so it carries named index labels
# first, second, third, and fourth.... instead of numeric ones like 0,1,2,3..
# Pandas can't align those labels to the existing numeric index.
# so therefore column 4 fills with NaN across all rows.
```

```python
# Add a new row at index 4 using loc

df1.loc[4] = [5, 9, 12, 0]
df1
```

## Insights and how it differs from sakila

- North wind has 500 rows and 4 columns, with a mix of numeric and text data. Min, max, and quartiles... unlike Sakila where all columns were text, the Unitprice ranges from '$2.00' to '$210.80' with a mean of -$23.37, suggesting most products are low-to-mid priced with a few expensive items pulling the max up.
- Quantity ordered averages about 24 units per line item, ranging from 1 to 120.
- OrderID should not be interpreted as a meaningful number, its mean and std are shown by describe() but they have no real value since its just an ID.
- Sakila's describe() only showed count, unique, top, and freq because every column was text. Northwind's describe() shows full numeric stats because it has actual quantitive columns worth mearsuing like Unitprice and Quantity.
