## Generating Random Numbers

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

## Generating Random Choices from an Array

```python
students = np.array([
    'Alice', 'Bob','Carlos','Diana','Eve','Frank','Grace', 'Hector', 'Isabel', 'James', 'Karen', 'Luis'
])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4), replace=False)
print(breakout_groups)
# Using replace false with 12 students filing a 3 by 4 grid means each student appears once
# no one is left out and no one is repeated. Its an effective way to randomly assign breakout groups
# as long as the class size matches the total number of slots in the grid
```

```python
# using probability values (p) for each of the students, sum to 1.0
# anything with a 0 will never be called on
p_values = [0.30, 0.20, 0.15, 0.10, 0.08, 0.07, 0.04, 0.03, 0.02, 0.01, 0.00, 0.00]
call_on_weighted = np.random.choice(students, size=5, p=p_values)
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
