## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.random(size=(4, 5))
print('r_1:', r_1)
print('r_2:\n', r_2)
print('r_3:\n', r_3)
print('r_4:\n', r_4)
print('r_5:\n', r_5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The elements were shuffled with np.random.shuffle() on a 2-D array. The rows themselves are rearranged. With the rows changing, the values within those rows stay together in the same order. Individual elements are not scattered across rows, the rows are swapped as units because shuffle operates along axis=0 by default.

```python
new_shuffle = np.random.permutation(r_4)
print('new_shuffle:\n', new_shuffle)
print('\nr_4 (original, unchanged):\n', r_4)
```

"When to use shuffle and permutation": 
Shuffle: You'll use this when you want to rearrange the original array in place and don't need the original order anymore.
Permutation: You'll use this when you want a new shuffled copy while leaving the original untouched. A safer route when you still need the original for comparison or for calculations.

## Mathematical Operations

```python
print('r_1:', r_1, ' shape:', r_1.shape)
print('r_2:\n', r_2, '  shape:', r_2.shape)
result_12 = np.add(r_1, r_2)
print('r_1 + r_2:\n', result_12, '  shape:', result_12.shape)
```

```python
print('r_1:', r_1, '  shape:', r_1.shape)
print('r_3:\n', r_3, '  shape:', r_3.shape)
result_13 = np.add(r_1, r_3)
print('r_1 + r_3:\n', result_13, '  shape:', result_13.shape)
```

```python
print('r_2:\n', r_2, '  shape:', r_2.shape)
print('r_4:\n', r_4, '  shape:', r_4.shape)
try:
    result_24 = np.add(r_2, r_4)
    print('r_2 + r_4:\n', result_24, '  shape:', result_24.shape)
except ValueError as e:
    print('Error:', e)
```

```python
# r_1 shape vs r_4 shape
try:
    print(np.add(r_1, r_4))
except ValueError as e:
    print('r_1 + r_4 Error:', e)

# r_2 shape vs r_4 shape
try:
    print(np.add(r_2, r_4))
except ValueError as e:
    print('r_2 + r_4 Error:', e)

# r_3 shape vs r_4 shape
try:
    print(np.add(r_3, r_4))
except ValueError as e:
    print('r_3 + r_4 Error:', e)
```

```python
# what happens with np.array([1]) to r_4 (broadcasts to (1,1) and then (4,5)
print(np.add(np.array([1]), r_4))
```

```python
# what happens with np.array([[1]]), r_4 (broadcasts to match (4,5) perfectly)
print(np.add(np.array([[1]]), r_4))
```

```python
# what happens with np.array([1,1,1,1,1]) to r_4 (treated as (1,5) then broadcasts to (4,5))
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))
```

```python
try:
    print(np.add(np.array([[1], [1]]), r_4))
except ValueError as e:
    # The error could not be broadcasted together with shapes (2,1) and (4,5)
    # Where the first dimension is 2 but r_4's dimension is 4.
    # But since niether is 1, that makes the broadcast fail
    print('Error:', e)
```

### How does broadcasting work?
NumPy compares the shapes from the compatible inward dimension. If two dimensions are compatible then they are equal or if one of them is 1. The result takes the larger size in each of the dimensions for example shape (5,) is treated as (1,5) due to np.array([1,1,1,1,1]). It'll then broadcast across all 4 rows of "r_4" (4,5) by adding 1 to every element. If any of the dimensions is incompatible or not equal to 1 then it'll raise a ValueError.

## Grocery Problem

```python
# Bag1: 2 Vegetables, 3 fruits, 4 snacks, 0 beverages
# Bag2: 4 vegetables, 0 fruits, 1 snack, 2 beverages
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print('bag1:', bag1)
print('bag2:', bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)
print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
category_totals = bag1 + bag2
print(f'''
There are {category_totals[0]} vegetables, {category_totals[1]} fruits, {category_totals[2]} boxes of snacks, and {category_totals[3]} beverages in the two grocery bags.
''')
```

## Teamwork

```python
# Ali knocked on: Monday 5, Tuesday 4, Wedsnesday 6, Thursday 4, Friday 2
ali = np.array([5, 4, 6, 4, 2])
print('Ali:', ali)
```

```python
# Ben missed Monday so Monday 0, then Tuesday 5, Weds 5, Thursday 3, Friday 4
ben = np.array([0, 5, 5, 3, 4])
print('Ben:', ben)
```

```python
# Carla knocked on: Monday 3, Tuesday 6, missed Weds 0, Thursday 3, Friday 4
carla = np.array([3, 6, 0, 3, 4])
print('Carla:', carla)
```

```python
# Each row is one person, each column is one day (mon-fri)
all_knocks = np.array([ali, ben, carla])
print('all_knocks:')
print(all_knocks)
```

```python
# Sum along axis=1
per_person = np.sum(all_knocks, axis=1)
print(f'''
Ali knocked on {per_person[0]} doors this week.
Ben knocked on {per_person[1]} doors this week.
Carla knocked on {per_person[2]} door this week.
''')

total_collective = np.sum(all_knocks)
print(f"The group collectively knocked on {total_collective} doors.")
```

```python
# 1. Sum across all people per day with axis=0
daily_totals = np.sum(all_knocks, axis=0)

# 2. Cumulative (rolling) total of door-knocks day by day
rolling_daily = np.cumsum(daily_totals)

# 3. Every 2 door-knocks = 1 estimated vote, so multiply rolling by 0.5
estimated_votes = rolling_daily * 0.5

print("Rolling total of estimated votes turned out, Mon through Fri:")
print(estimated_votes)
```
