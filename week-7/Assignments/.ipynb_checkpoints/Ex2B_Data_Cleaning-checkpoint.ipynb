# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd

# recreating file (test_data.csv) (didn''t want to run)

test_data = {
    'ID':   [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
pd.DataFrame(test_data).to_csv('test_data.csv', index=False)

df = pd.read_csv('test_data.csv')
df.describe()
```

## Filtering a DataFrame

```python
df[df['Age'] > 18]
```

```python
df[(df['Score'] > 50) & (df['Score'] < 100)]
```

## Modifying a DataFrame

```python
# Dropping rows where Score is less than 0
# should make it invalid
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
# Dropping the city column
df_cleaned = df_cleaned.drop(columns='City')
df_cleaned
```

```python
# Fill missing Age values with the mean Age of the DataFrame
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
df_cleaned
```

```python
# Replace any score value of 0 with the mean score
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
```

```python
# Remove any remaining rows that still contain NaN values
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
# Set ID as the index labels, then drop the ID column
df_cleaned = df_cleaned.set_index('ID')
df_cleaned
```
