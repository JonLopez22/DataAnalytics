# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating x and y data

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
# Add markers
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
# Dashed red line
plt.plot(x, y, linestyle='--', color='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# square green marker, size 10
plt.plot(x, y, linestyle='--', color='red', marker='s', markersize=10, markerfacecolor='green')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Two additional marker styles

# triangle, blue
plt.plot(x, y, marker='^', color='blue', linestyle='-')
plt.title('Triangle Marker - Blue')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()

# Diamond, purple
plt.plot(x, y, marker='D', color='purple', linestyle='-.')
plt.title('Diamond Marker - Purple')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# bar chart basic
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python
# Bar chart customized
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Product Sales')
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.show()
```

```python
# Scatter plot (basic)
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python
# scatter plot customized with size and color map
x = np.random.rand(50)
y = np.random.rand(50)
s = np.random.rand(50) * 100

plt.scatter(x, y, s=s, c=y, cmap='viridis')
plt.colorbar(label='Y value')
plt.title('Scatter Plot with Size and Color')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## Subplots

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Subplot 1: Sine wave
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax1.plot(x, y, color='steelblue')
ax1.set_title('Sine Wave')
ax1.set_xlabel('X')
ax1.set_ylabel('sin(X)')

# Subplot 2: Histogram
data = np.random.randn(1000)
ax2.hist(data, bins=30, color='coral', edgecolor='black')
ax2.set_title('Histogram of Random Numbers')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
