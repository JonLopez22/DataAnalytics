# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Creating the dataframe
data = {'Product':    ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],   # 9 items
        'Region':     ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],  # 9 items
        'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],               # 9 items
        'Price':      [100, 200, 150, 100, 200, 150, 100, 200, 150]}  
df_sales = pd.DataFrame(data)
df_sales
```

```python
# Adding a revenue column by multiplying Units_Sold and Price for each row
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
# Calculate total Units_Sold and total Revenue across all products and regions
df_sales[['Units_Sold', 'Revenue']].sum()
```

```python
# Caluclate average Units_Sold and average Revenue across all rows
df_sales[['Units_Sold', 'Revenue']].mean()
```

## Grouping Data

```python
# Group by Product and sum Units_Sold and Revenue for each product
df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
```

```python
# Group by Region and caluclate average Units_Sold and Revenue for each region
df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
```

```python
# Group by both Product and Region, then sum Units_Sold and Revenue for each combination 
df_sales.groupby(['Product', 'Region',])[['Units_Sold', 'Revenue']].sum()
```

## Sorting and Ranking Data

```python
# Sort df_sales by Units_Sold in ascending order
df_sales.sort_values('Units_Sold')
```

```python
# Sort df_sales by Revenue in descending order
df_sales.sort_values('Revenue', ascending=False)
```

```python
# Add a Revenue_Rank column with rank 1 being the highest revenue
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False).astype(int)
df_sales # review sales with the new Revenue_Rank column
```

## Merging DataFrames

```python
# Create df_products mapping each Product to its Category
product_data = {'Product': ['A', 'B', 'C'],
                'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)
df_products
```

```python
# Merge df_sales and df_products on the shared Product column
df_merged = pd.merge(df_sales, df_products, on='Product')
# This is an inner join by default so only the matching products are kept
df_merged
```

```python
# Group df_merged by Category and sum Revenue to see which category earns the most 
df_merged.groupby('Category')['Revenue'].sum()
```
